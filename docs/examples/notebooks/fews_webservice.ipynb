```python
from pydantic_ai import format_as_xml
from setup_quarto import setup_dot_env_docs

ctx = setup_dot_env_docs()
```

## Get Locations{#sec-get-locations}
De webservice van FEWS heeft meerdere manieren om data te filteren. We gebruiken nu locaties en parameters om de data te filteren. De locaties zijn de meetpunten waar de data vandaan komt. We kunnen deze ophalen met de functie `get_locations`:

```python
filter_locations_name = "Heino"
```

```python
from talk_with_fews.tools import get_locations
from talk_with_fews.models import GetLocationInput, GetLocationOutput

get_location_input = GetLocationInput(
    number_of_locations=2, filter_locations_name=filter_locations_name
)

get_location_output: GetLocationOutput = await get_locations(
    ctx=ctx, input=get_location_input
)
```

```python
print(format_as_xml(get_location_output, root_tag="GetLocationOutput"))
```

## Get Parameters{#sec-get-parameters}
De parameters zijn gerelateerd aan het type data. Deze functie heeft een parameter `filter_parameters_name` waarmee we kunnen filteren op de naam van de parameter. We kunnen ook het aantal parameters beperken met de parameter `number_of_parameters`.


```python
filter_parameters_name = "E"
```

```python
from talk_with_fews.tools import get_parameters
from talk_with_fews.models import GetParameterInput, GetParameterOutput

get_parameter_input = GetParameterInput(
    number_of_parameters=2, filter_parameters_name=filter_parameters_name
)
get_parameter_output: GetParameterOutput = await get_parameters(
    ctx=ctx, input=get_parameter_input
)
```

```python
print(format_as_xml(get_parameter_output, root_tag="GetParameterOutput"))
```

## Get Series Headers{#sec-get-series-headers}
De headers van de series zijn de metadata. Hiermee kunnen we een koppeling maken tussen locaties en parameter. 
Eén locatie of één parameter wordt opgegeven om hiervoor de bijbehorende metadata op te halen.
We geven voor beide een voorbeeld:


```python
location_id = get_location_output.locations[0].locationId
location_id
```

```python
from talk_with_fews.tools import get_series_headers
from talk_with_fews.models import GetSeriesHeadersInput, GetSeriesHeadersOutput

get_series_headers_input = GetSeriesHeadersInput(location_id=location_id)
get_series_headers_output: GetSeriesHeadersOutput = await get_series_headers(
    ctx=ctx, input=get_series_headers_input
)
print(format_as_xml(get_series_headers_output, root_tag="GetSeriesHeadersOutput"))
```

```python
parameter_id = get_parameter_output.parameters[0].id
parameter_id
```

```python
get_series_headers_input = GetSeriesHeadersInput(parameter_id=parameter_id)
get_series_headers_output: GetSeriesHeadersOutput = await get_series_headers(
    ctx=ctx, input=get_series_headers_input
)
print(format_as_xml(get_series_headers_output, root_tag="GetSeriesHeadersOutput"))
```

## Execute Query Webservice {#sec-execute-query-webservice}
Bij het ophalen van de parameters, locaties en series headers hebben we de metadata opgehaald. Deze slaan wij in een cache database op om deze later te kunnen gebruiken, met als voordeel dat we niet steeds dezelfde data hoeven op te halen, maar ook dat we alle vrijheid hebben om de data te filteren en te combineren.
We kunnen nu de meta data verder analyseren, bijvoorbeeld ook met een ruimtelijke query:

```python
from talk_with_fews.tools import (
    query_webservice_cache,
)
from talk_with_fews.models import (
    SQLQueryInput,
    SQLQueryOutput,
)

sql_query_input = SQLQueryInput(
    query="SELECT locationId, shortName, lat, lon, ST_Distance(geom, ST_Point(5.67, 52.035)) as distance_degrees FROM locations WHERE ST_DWithin(geom, ST_Point(5.67, 52.035), 0.1) ORDER BY distance_degrees LIMIT 30",
)
sql_query_output: SQLQueryOutput = await query_webservice_cache(
    ctx=ctx, input=sql_query_input
)
```

```python
print(format_as_xml(sql_query_output, root_tag="SQLQueryOutput"))
```

## Get Series Statistics{#sec-get-series-statistics}
Met de meta data die we hebben opgehaald, kunnen we nu de series statistieken ophalen. Dit is iets sneller dan de data zelf ophalen, en scheelt data overdracht. Hier moet een parameter en locatie id worden opgegeven, waar de data van wordt opgehaald. Daarnaast ook de tijden waarover de statistieken moeten worden berekend.

```python
location_id = get_series_headers_output.available_ids["locations"][1]
parameter_id, location_id
```

```python
from datetime import datetime, timezone


start_time = datetime(1990, 1, 1, tzinfo=timezone.utc)
end_time = datetime(2025, 1, 1, tzinfo=timezone.utc)
print(start_time, end_time)
```

```python
from talk_with_fews.tools import get_series_statistics
from talk_with_fews.models import (
    GetSeriesStatisticsInput,
    GetSeriesStatisticsOutput,
)

get_series_statistics_input = GetSeriesStatisticsInput(
    parameter_id=parameter_id,
    location_id=location_id,
    start_time=start_time,
    end_time=end_time,
)
get_series_statistics_output: GetSeriesStatisticsOutput = await get_series_statistics(
    ctx=ctx, input=get_series_statistics_input
)
```

```python
print(format_as_xml(get_series_statistics_output, root_tag="GetSeriesStatisticsOutput"))
```

## Get Series Data - Work in progress to stream
Naast de statistiek van de series, kunnen we ook de data zelf ophalen.
Hier moeten we ook een parameter en locatie id opgeven, maar ook een start en eind tijd. Deze kunnen we ophalen uit de statistieken die we eerder hebben opgehaald.

```python
from datetime import timedelta


start_time = get_series_statistics_output.series_statistics[0].first_value_time
end_time = start_time + timedelta(days=10)
print(start_time, end_time)
```

```python
parameter_id, location_id
```

```python
from talk_with_fews.tools import get_series_data
from talk_with_fews.models import (
    GetSeriesDataInput,
    GetSeriesDataOutput,
)

# Get the module_instance_id from the statistics output
module_instance_id = get_series_statistics_output.series_statistics[
    0
].module_instance_id

get_series_data_input = GetSeriesDataInput(
    parameter_id=parameter_id,
    location_id=location_id,
    module_instance_id=module_instance_id,
    start_time=start_time,
    end_time=end_time,
    include_flag_sources=False,
)

get_series_data_output: GetSeriesDataOutput = await get_series_data(
    ctx=ctx, input=get_series_data_input
)
```

```python
print(format_as_xml(get_series_data_output, root_tag="GetSeriesDataOutput"))
```
